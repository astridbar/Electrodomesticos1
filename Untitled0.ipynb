```python
import matplotlib.pyplot as plt
import numpy as np

# Definir las dimensiones (15 competencias)
categories = ["Cultura", "Liderazgo", "Presupuesto", "Innovación", "Tecnología",
              "Gestión de Datos", "Seguridad", "Habilidades", "Procesos", "Infraestructura",
              "Colaboración", "Automatización", "Capacidades", "Reclutamiento", "Aprendizaje"]
N = len(categories)

# Valores de evaluación (por ejemplo, todos en nivel 3, excepto Gestión de Proyectos en 4)
values = [3, 3, 3, 3, 3, 3, 3, 3, 4, 3, 3, 3, 3, 3, 3]

# Cerrar el gráfico agregando el primer valor al final
values += values[:1]

# Calcular los ángulos para cada dimensión (en radianes)
angles = [n / float(N) * 2 * np.pi for n in range(N)]
# Cerrar el círculo
angles += angles[:1]

# Crear el gráfico radar
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw={'polar': True})
ax.plot(angles, values, 'o-', linewidth=2, color='blue')
ax.fill(angles, values, alpha=0.25, color='blue')

# Configurar las etiquetas angulares: se usan los ángulos originales (sin el último duplicado)
ax.set_thetagrids(np.degrees(angles[:-1]), categories)
ax.set_ylim(0, 5)
ax.set_title("Evaluación de Madurez Digital de DIACO S.A.S.", y=1.08)

plt.tight_layout()
plt.savefig("madurez_digital_diaco.png", dpi=300)
plt.show()
```
